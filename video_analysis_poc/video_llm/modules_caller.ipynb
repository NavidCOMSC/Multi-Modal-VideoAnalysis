### Local Retriever for LiteLLM Module Selection

```python
import litellm
import json
import requests
```

```python
litellm.api_key = "sk-"
# Set the API base URL to the custom endpoint of UK-DS-IDA-06-005-litellm-url
litellm.api_base = "https://genai-lounge-nx-litellm-uat-emea.zurich.com"
```

```python
response = requests.get(f"{litellm.api_base}/models", headers={"Authorization": f"Bearer {litellm.api_key}"})
```

```python
models = response.json()
print(json.dumps(models, indent=2))
```

```python

with open("available_models.txt", "w") as f:
    f.write(json.dumps(models, indent=2))
print("Models saved to available_models.txt")
```

```python
# Inspect raw response first
print("Status:", response.status_code)
print("Content-Type:", response.headers.get("Content-Type"))
print("Body preview:", response.text[:500])

try:
	data = response.json()
	models = data.get("data", data) if isinstance(data, dict) else data
	print(json.dumps(models, indent=2))
except json.JSONDecodeError as e:
	print(f"Failed to parse JSON: {e}")
	print("Full response text:")
	print(response.text)
```
